```python
import os
import glob
from ultralytics import YOLO

def main():
    print("🚀 Starting YOLOv8 Model Training for Biomass Valuation (V2 Dataset)...")

    # 1. Dataset ka data.yaml dhoondhna
    print("🔍 Looking for dataset...")
    yaml_files = glob.glob('**/data.yaml', recursive=True)

    if not yaml_files:
        print("❌ Error: data.yaml not found! Please check if Roboflow downloaded the dataset correctly.")
        return

    data_path = os.path.abspath(yaml_files[0])
    print(f"✅ Found dataset configuration at: {data_path}")

    # 2. Load the YOLOv8 model (Small version is good for a balance of speed and accuracy)
    print("⏳ Loading YOLOv8 model...")
    model = YOLO('yolov8s.pt')

    # 3. Training configuration
    print("🔥 Starting training process...")
    results = model.train(
        data=data_path,
        epochs=150,          # Training cycles (Aap ise apne hardware ke hisaab se adjust kar sakti hain)
        imgsz=512,           # Naye dataset (Stretch to 512x512) ke sath sync karne ke liye
        batch=16,            # Batch size
        name='stubble_v2',   # Naye model ke folder ka naam
        patience=25          # Agar 25 epochs tak model improve nahi hota, toh training auto-stop ho jayegi
    )

    print("\n✅ Training Complete!")
    print("Your best model is saved in the 'runs/detect/stubble_v2/weights' folder as 'best.pt'.")

if __name__ == '__main__':
    main()
```
